```python
# vid 4
"""
given an array of ints, find the pair of adjacent elements
that have the largest product and return that product

ex:
inputArray = [3, 6, -2, -5, 7, 3]
output = 21

"""
def solution(inputArray):
    product = inputArray[0] * inputArray[1]

    for i in range(len(inputArray) - 1):
        prod = inputArray[i] * inputArray[i+1]
        if prod > product:
            product = prod

    return product


# multiple inputs for test cases
inputArray = [3, 6, -2, -5, 7, 3]

twoInts = [1, -2]

biggerSumFirst = [7, 3, 6, -2, -5]
print(solution(biggerSumFirst))
```
